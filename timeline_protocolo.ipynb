```python
import matplotlib.pyplot as plt
%matplotlib qt
```

```python

# Datos actualizados
etapas = [
    "Explicación \n  Formulario de Consentimiento Informado", 
    "Evaluaciones Iniciales \n (catastrofización, sensibilidad, \n rigidez muscular, PPT )", 
    "Familiarización \n Umbrales de Cuchilla", 
    "Descanso", 
    "Repetición Evaluaciones", 
    "Descanso", 
    "CPM",
    "Umbrales de Cuchilla"
]
duraciones = [15, 15, 15, 10, 15, 15, 5, 10]

# Colores (reciclando y añadiendo uno para la nueva etapa)
colores = ['#4a90e2', '#50e3c2', '#f5a623', '#d3d3d3', '#50e3c2', '#d3d3d3', '#e15759', '#f5a623']

# Calcular inicios
inicios = [0]
for i in range(len(duraciones) - 1):
    inicios.append(inicios[-1] + duraciones[i])

fig, ax = plt.subplots(figsize=(14, 4))

# Dibujar la barra única
y_pos = 0
altura = 1

for i, (etapa, duracion, inicio, color) in enumerate(zip(etapas, duraciones, inicios, colores)):
    ax.barh(y_pos, duracion, left=inicio, height=altura, color=color, edgecolor='white')
    
    # Texto de duración en el centro de la barra
    center_x = inicio + duracion / 2
    ax.text(center_x, y_pos, f"{duracion}'", ha='center', va='center', color='black', fontsize=9, fontweight='bold')
    
    # Etiqueta de la etapa (alternando arriba y abajo para que no se pisen)
    # Offset vertical
    if i % 2 == 0:
        y_text = y_pos + altura/2 + 0.1
        va = 'bottom'
    else:
        y_text = y_pos - altura/2 - 0.1
        va = 'top'
        
    ax.text(center_x, y_text, etapa, ha='center', va=va, fontsize=10, rotation=0, wrap=True)

# Configuración de ejes
ax.set_ylim(-1.5, 1.5)
ax.set_xlim(0, sum(duraciones))
ax.set_yticks([])
ax.set_xlabel('Tiempo Total (Minutos)', fontsize=12)
ax.set_title('Línea de Tiempo Lineal de la Sesión', fontsize=14)

# Grid en X
ax.set_xticks(range(0, sum(duraciones) + 5, 10))
ax.grid(axis='x', linestyle='--', alpha=0.5)

# Quitar bordes innecesarios
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(True)

plt.tight_layout()
#plt.savefig('timeline_lineal.png', dpi=300)
```
